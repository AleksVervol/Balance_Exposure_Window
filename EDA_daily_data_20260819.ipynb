```python
import pandas as pd
import yfinance as yf
```

```python
crypto_daily = yf.download(
    ["BTC-USD", "ETH-USD"],
    start="2020-01-01",
    end="2026-08-20",
    interval="1d",
    auto_adjust=False
)

crypto_daily.head()
```

```python
crypto_daily.to_csv("data/raw/daily/crypto_daily_full.csv")
```

```python
crypto_daily.info()
```

```python
crypto_daily.shape
```

```python
crypto_daily.isna().sum()
```

```python
btc_hourly = yf.download(
    "BTC-USD",
    start="2022-05-01",
    end="2022-05-25",
    interval="1h",
    auto_adjust=False
)

btc_hourly.head()
```

```python
#Terra/Luna - 2022-05-05/13
#Crypto-native
#Asset/stablecoin collapse

crypto_may_2022 = crypto_daily.loc['2022-05-01':'2022-05-31']

crypto_may_2022
```

```python
crypto_may_2022.isna().sum()
```

```python
crypto_may_2022["Close"].plot(
    figsize=(12, 5),
    title="BTC and ETH Price — May 2022"
)
```

```python
crypto_may_2022["Close"]["BTC-USD"].plot(
    figsize=(12, 5),
    title="BTC Price — May 2022"
)
```

```python
crypto_may_2022["Close"]["ETH-USD"].plot(
    figsize=(12, 5),
    title="ETH Price — May 2022"
)
```

```python
crypto_may_2022["BTC_pct_change"] = crypto_may_2022["Close"]["BTC-USD"].pct_change() * 100
crypto_may_2022["ETH_pct_change"] = crypto_may_2022["Close"]["ETH-USD"].pct_change() * 100
crypto_may_2022.head()
```

```python
#Terra/Luna - 2022-05-05/13
crypto_may_2022[["BTC_pct_change", "ETH_pct_change"]].sort_values("BTC_pct_change").head()
```

```python
crypto_may_2022[["BTC_pct_change", "ETH_pct_change"]].sort_values("ETH_pct_change").head()
```
